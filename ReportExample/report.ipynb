This notebook demonstrates basic matplotlib plots and how to display a local image.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Line Plot

```python
x = np.linspace(0, 2 * np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y_sin, label="sin(x)")
ax.plot(x, y_cos, label="cos(x)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Sine and Cosine")
ax.legend()
plt.tight_layout()
plt.show()
```

## Scatter Plot

```python
rng = np.random.default_rng(42)
x = rng.normal(size=200)
y = 0.5 * x + rng.normal(scale=0.4, size=200)

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(x, y, alpha=0.6, edgecolors="k", linewidths=0.5)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Random Scatter Plot")
plt.tight_layout()
plt.show()
```
